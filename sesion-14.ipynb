## Teorema CAP

En sistemas distribuidos solo puedes garantizar 2 de 3:

| Sistema | Consistency | Availability | Partition | Ejemplo |
|---------|-------------|--------------|-----------|----------|
| **CP** | ✅ | ❌ | ✅ | MongoDB, HBase |
| **AP** | ❌ | ✅ | ✅ | Cassandra, DynamoDB |
| **CA** | ✅ | ✅ | ❌ | RDBMS tradicional |

## Tipos de NoSQL

```python
# Simulación de diferentes modelos de datos NoSQL

# ========================================
# 1. KEY-VALUE (como Redis)
# ========================================
print("=== KEY-VALUE ===")

key_value_store = {}

# Operaciones típicas
key_value_store["user:1234"] = {"name": "Ana", "email": "ana@example.com"}
key_value_store["session:abc"] = {"user_id": 1234, "expires": "2025-01-01"}
key_value_store["cart:1234"] = ["producto_1", "producto_2"]

print(f"GET user:1234 -> {key_value_store['user:1234']}")
print(f"GET session:abc -> {key_value_store['session:abc']}")
print("\nUso: Cache, sesiones, configuración")
```

```python
# ========================================
# 2. DOCUMENT (como MongoDB)
# ========================================
print("=== DOCUMENT STORE ===")

# Colección de documentos
productos = [
    {
        "_id": "prod_001",
        "nombre": "Laptop Pro",
        "precio": 999.99,
        "categorias": ["electrónica", "ordenadores"],
        "specs": {
            "cpu": "Intel i7",
            "ram": "16GB",
            "storage": "512GB SSD"
        },
        "reviews": [
            {"user": "user_1", "rating": 5, "comment": "Excelente"},
            {"user": "user_2", "rating": 4, "comment": "Muy bueno"}
        ]
    },
    {
        "_id": "prod_002",
        "nombre": "Mouse Wireless",
        "precio": 29.99,
        "categorias": ["electrónica", "accesorios"]
        # Nota: No tiene specs ni reviews - schema flexible
    }
]

# Query simulado
print(f"Documento 1: {productos[0]['nombre']}")
print(f"  Specs: {productos[0]['specs']}")
print(f"  Reviews: {len(productos[0]['reviews'])} opiniones")
print(f"\nDocumento 2: {productos[1]['nombre']}")
print(f"  Sin specs definidas (schema flexible)")
print("\nUso: Catálogos, CMS, perfiles de usuario")
```

```python
# ========================================
# 3. COLUMNAR (como Cassandra)
# ========================================
import pandas as pd

print("=== COLUMNAR STORE ===")

# Modelo wide-column
# Row key: user_id
# Column families: profile, activity, preferences

data = {
    'user_001': {
        'profile:name': 'Ana García',
        'profile:email': 'ana@mail.com',
        'activity:last_login': '2024-01-15',
        'activity:page_views': 150,
        'preferences:theme': 'dark',
        'preferences:language': 'es'
    },
    'user_002': {
        'profile:name': 'Luis Pérez',
        'profile:email': 'luis@mail.com',
        'activity:last_login': '2024-01-14'
        # Sin preferences - sparse
    }
}

print("Row: user_001")
for col, val in data['user_001'].items():
    print(f"  {col}: {val}")

print("\nRow: user_002 (sparse - menos columnas)")
for col, val in data['user_002'].items():
    print(f"  {col}: {val}")

print("\nUso: Time series, logs, IoT, analytics")
```

```python
# ========================================
# 4. GRAPH (como Neo4j)
# ========================================
print("=== GRAPH DATABASE ===")

# Representación simplificada de grafo
nodes = {
    'user_1': {'type': 'User', 'name': 'Ana'},
    'user_2': {'type': 'User', 'name': 'Luis'},
    'user_3': {'type': 'User', 'name': 'María'},
    'prod_1': {'type': 'Product', 'name': 'Laptop'},
    'prod_2': {'type': 'Product', 'name': 'Mouse'}
}

edges = [
    ('user_1', 'FOLLOWS', 'user_2'),
    ('user_2', 'FOLLOWS', 'user_3'),
    ('user_1', 'PURCHASED', 'prod_1'),
    ('user_2', 'PURCHASED', 'prod_1'),
    ('user_2', 'PURCHASED', 'prod_2'),
    ('user_3', 'REVIEWED', 'prod_1')
]

print("Nodos:")
for node_id, props in nodes.items():
    print(f"  ({node_id}:{props['type']} {{name: '{props['name']}'}})")

print("\nRelaciones:")
for src, rel, dst in edges:
    print(f"  ({src})-[:{rel}]->({dst})")

# Query ejemplo: ¿Qué compraron los amigos de Ana?
print("\nQuery: Productos comprados por seguidos de user_1")
amigos_ana = [e[2] for e in edges if e[0] == 'user_1' and e[1] == 'FOLLOWS']
productos_amigos = [e[2] for e in edges if e[0] in amigos_ana and e[1] == 'PURCHASED']
print(f"  Resultado: {productos_amigos}")

print("\nUso: Redes sociales, recomendaciones, detección de fraude")
```

## Data Lake vs Data Warehouse

| Aspecto | Data Warehouse | Data Lake |
|---------|---------------|----------|
| Schema | Schema-on-write | Schema-on-read |
| Datos | Estructurados | Cualquier tipo |
| Procesamiento | SQL, BI | ML, Analytics |
| Usuarios | Analistas negocio | Data Scientists |
| Coste | Alto (storage optimizado) | Bajo (object storage) |
