## Ejercicio de repaso 1: Complejidad

```python
# ¿Cuál es la complejidad de estas operaciones?

# Operación 1: Acceder a un elemento por índice
lista = [1, 2, 3, 4, 5]
x = lista[2]  # Complejidad: O(1)

# Operación 2: Buscar un elemento en una lista
encontrado = 3 in lista  # Complejidad: O(n)

# Operación 3: Buscar en un diccionario
diccionario = {"a": 1, "b": 2, "c": 3}
encontrado = "b" in diccionario  # Complejidad: O(1)

# Operación 4: Ordenar una lista
ordenada = sorted(lista)  # Complejidad: O(n log n)

# Operación 5: Doble bucle anidado
for i in lista:
    for j in lista:
        pass  # Complejidad: O(n²)

print("=== RESPUESTAS ===")
print("1. Acceso por índice: O(1)")
print("2. Búsqueda en lista: O(n)")
print("3. Búsqueda en dict: O(1)")
print("4. Ordenación: O(n log n)")
print("5. Doble bucle: O(n²)")
```
