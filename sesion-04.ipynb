## 1. Operaciones con Pandas

```python
import pandas as pd
import numpy as np

# ========================================
# CREAR DATAFRAME DE EJEMPLO
# ========================================
df = pd.DataFrame({
    'nombre': ['Ana', 'Luis', 'María', 'Pedro', 'Laura', 'Carlos'],
    'edad': [25, 30, 22, 35, 28, 40],
    'ciudad': ['Madrid', 'Barcelona', 'Sevilla', 'Madrid', 'Valencia', 'Barcelona'],
    'salario': [35000, 42000, 28000, 55000, 38000, 60000],
    'fecha_ingreso': ['2020-01-15', '2019-06-01', '2021-03-10', '2018-09-20', '2022-01-05', '2017-04-15']
})

print("=== DATAFRAME ORIGINAL ===")
print(df)

# ========================================
# OPERACIONES DE EXPLORACIÓN
# ========================================
print("\n=== EXPLORACIÓN ===")
print(f"Dimensiones: {df.shape}")
print(f"Columnas: {df.columns.tolist()}")
print(f"\nTipos de datos:\n{df.dtypes}")
print(f"\nEstadísticas:\n{df.describe()}")
```

```python
# ========================================
# SELECCIÓN Y FILTRADO
# ========================================
print("=== SELECCIÓN Y FILTRADO ===")

# Filtrar filas
print("\nPersonas con edad > 25:")
print(df[df['edad'] > 25])

print("\nPersonas de Madrid con salario > 40000:")
print(df[(df['ciudad'] == 'Madrid') & (df['salario'] > 40000)])
```

```python
# ========================================
# TRANSFORMACIÓN
# ========================================
print("=== TRANSFORMACIÓN ===")

# Añadir columna calculada
df['salario_mensual'] = df['salario'] / 12
print("\nCon salario mensual:")
print(df[['nombre', 'salario', 'salario_mensual']])

# Categorizar
df['categoria_edad'] = df['edad'].apply(lambda x: 'joven' if x < 30 else 'senior')
print("\nCon categoría de edad:")
print(df[['nombre', 'edad', 'categoria_edad']])
```

```python
# ========================================
# AGREGACIÓN
# ========================================
print("=== AGREGACIÓN ===")

print("\nSalario medio por ciudad:")
print(df.groupby('ciudad')['salario'].mean())

print("\nEstadísticas por ciudad:")
print(df.groupby('ciudad').agg({
    'salario': ['mean', 'min', 'max'],
    'edad': 'mean'
}))
```

```python
# ========================================
# COMBINACIÓN (JOIN)
# ========================================
print("=== COMBINACIÓN (JOIN) ===")

# Crear DataFrame adicional
df_departamentos = pd.DataFrame({
    'ciudad': ['Madrid', 'Barcelona', 'Sevilla', 'Valencia'],
    'departamento': ['Central', 'Noreste', 'Sur', 'Este'],
    'presupuesto': [100000, 80000, 60000, 70000]
})

print("DataFrame de departamentos:")
print(df_departamentos)

# Merge (JOIN)
df_completo = df.merge(df_departamentos, on='ciudad', how='left')
print("\nDatos combinados:")
print(df_completo[['nombre', 'ciudad', 'departamento', 'presupuesto']])
```

## 2. Extracción de datos de API REST

```python
import requests

# API pública de ejemplo
url = "https://jsonplaceholder.typicode.com/users"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    
    usuarios = response.json()
    
    print(f"Status: {response.status_code}")
    print(f"Usuarios obtenidos: {len(usuarios)}")
    
    # Convertir a DataFrame
    df_usuarios = pd.json_normalize(usuarios)
    print("\nColumnas disponibles:")
    print(df_usuarios.columns.tolist())
    
    print("\nPrimeros usuarios:")
    print(df_usuarios[['name', 'email', 'address.city']].head())
    
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
```
