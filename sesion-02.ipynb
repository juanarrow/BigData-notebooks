## 2. Ejercicio: Comparación de rendimiento

```python
# Comparación de rendimiento de estructuras de datos
import time
import random
from collections import deque

# Datos de prueba
n = 100000
datos = [random.randint(1, 1000000) for _ in range(n)]
buscar = random.sample(datos, 1000)

# 1. Lista
print("=== LISTA ===")
lista = datos.copy()
inicio = time.time()
encontrados = [x for x in buscar if x in lista]
print(f"Búsqueda en lista: {time.time() - inicio:.4f}s")

# 2. Conjunto (Set)
print("\n=== CONJUNTO ===")
conjunto = set(datos)
inicio = time.time()
encontrados = [x for x in buscar if x in conjunto]
print(f"Búsqueda en conjunto: {time.time() - inicio:.4f}s")

# 3. Diccionario
print("\n=== DICCIONARIO ===")
diccionario = {x: True for x in datos}
inicio = time.time()
encontrados = [x for x in buscar if x in diccionario]
print(f"Búsqueda en diccionario: {time.time() - inicio:.4f}s")

# Conclusión
print("\n¿Por qué hay tanta diferencia de tiempos?")
print("Lista: O(n) - debe recorrer todos los elementos")
print("Set/Dict: O(1) - usa hash table para acceso directo")
```

## 3. Implementación de Pila (Stack)

```python
class Pila:
    """Implementación de una pila (LIFO)"""
    
    def __init__(self):
        self.items = []
    
    def push(self, item):
        """Añadir elemento al tope - O(1)"""
        self.items.append(item)
    
    def pop(self):
        """Quitar y devolver el tope - O(1)"""
        if not self.is_empty():
            return self.items.pop()
        raise IndexError("Pila vacía")
    
    def peek(self):
        """Ver el tope sin quitarlo - O(1)"""
        if not self.is_empty():
            return self.items[-1]
        raise IndexError("Pila vacía")
    
    def is_empty(self):
        return len(self.items) == 0
    
    def size(self):
        return len(self.items)

# Ejemplo de uso
pila = Pila()
pila.push(10)
pila.push(20)
pila.push(30)

print(f"Tope: {pila.peek()}")  # 30
print(f"Pop: {pila.pop()}")    # 30
print(f"Pop: {pila.pop()}")    # 20
print(f"Tamaño: {pila.size()}") # 1
```

## 4. Implementación de Cola (Queue)

```python
from collections import deque

class Cola:
    """Implementación de una cola (FIFO) usando deque"""
    
    def __init__(self):
        self.items = deque()
    
    def enqueue(self, item):
        """Añadir al final - O(1)"""
        self.items.append(item)
    
    def dequeue(self):
        """Quitar del frente - O(1)"""
        if not self.is_empty():
            return self.items.popleft()
        raise IndexError("Cola vacía")
    
    def front(self):
        """Ver el frente sin quitarlo - O(1)"""
        if not self.is_empty():
            return self.items[0]
        raise IndexError("Cola vacía")
    
    def is_empty(self):
        return len(self.items) == 0
    
    def size(self):
        return len(self.items)

# Ejemplo de uso
cola = Cola()
cola.enqueue("Cliente 1")
cola.enqueue("Cliente 2")
cola.enqueue("Cliente 3")

print(f"Frente: {cola.front()}")     # Cliente 1
print(f"Dequeue: {cola.dequeue()}")  # Cliente 1
print(f"Dequeue: {cola.dequeue()}")  # Cliente 2
print(f"Tamaño: {cola.size()}")      # 1
```

## 5. Comparación: Lista vs Deque

```python
import time
from collections import deque

n = 100000

# Inserción al inicio - Lista
lista = []
inicio = time.time()
for i in range(n):
    lista.insert(0, i)  # O(n) por cada inserción
tiempo_lista = time.time() - inicio
print(f"Lista - insertar al inicio: {tiempo_lista:.4f}s")

# Inserción al inicio - Deque
d = deque()
inicio = time.time()
for i in range(n):
    d.appendleft(i)  # O(1) por cada inserción
tiempo_deque = time.time() - inicio
print(f"Deque - insertar al inicio: {tiempo_deque:.4f}s")

print(f"\nDeque es {tiempo_lista/tiempo_deque:.1f}x más rápido para inserción al inicio")
```
