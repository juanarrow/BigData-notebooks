## 1. Lógica Proposicional en filtrado de datos

```python
import pandas as pd

# Crear DataFrame de ejemplo
df = pd.DataFrame({
    'nombre': ['Ana', 'Luis', 'María', 'Pedro', 'Laura'],
    'edad': [25, 17, 30, 22, 16],
    'ciudad': ['Málaga', 'Sevilla', 'Málaga', 'Granada', 'Málaga'],
    'categoria': ['A', 'B', 'A', 'C', 'B'],
    'eliminado': [False, False, True, False, False]
})

print("DataFrame original:")
print(df)

# Filtro con lógica AND: Mayores de 18 Y de Málaga
print("\nFiltro AND (edad > 18 Y ciudad == 'Málaga'):")
resultado_and = df[(df['edad'] > 18) & (df['ciudad'] == 'Málaga')]
print(resultado_and)

# Filtro con lógica OR: Categoría A O categoría B
print("\nFiltro OR (categoria == 'A' O categoria == 'B'):")
resultado_or = df[(df['categoria'] == 'A') | (df['categoria'] == 'B')]
print(resultado_or)

# Filtro con NOT: No eliminados
print("\nFiltro NOT (no eliminados):")
resultado_not = df[~df['eliminado']]
print(resultado_not)
```

## 2. Clases de Complejidad

| Clase | Descripción | Ejemplos |
|-------|-------------|----------|
| **P** | Resoluble en tiempo polinomial | Ordenación, Búsqueda binaria |
| **NP** | Verificable en tiempo polinomial | Factorización, SAT |
| **NP-Completo** | Los más difíciles de NP | TSP, Mochila, Coloreo de grafos |

## 3. Impacto de la complejidad en Big Data

```python
import time
import matplotlib.pyplot as plt

def medir_complejidad(funcion, tamanos):
    tiempos = []
    for n in tamanos:
        datos = list(range(n))
        inicio = time.time()
        funcion(datos)
        tiempos.append(time.time() - inicio)
    return tiempos

# O(n) - Recorrido lineal
def lineal(datos):
    suma = 0
    for x in datos:
        suma += x
    return suma

# O(n²) - Doble bucle
def cuadratico(datos):
    contador = 0
    for i in datos:
        for j in datos:
            contador += 1
    return contador

# O(n log n) - Ordenación
def nlogn(datos):
    return sorted(datos)

# Comparar
tamanos = [100, 500, 1000, 2000, 3000, 4000, 5000]

t_lineal = medir_complejidad(lineal, tamanos)
t_nlogn = medir_complejidad(nlogn, tamanos)
t_cuadratico = medir_complejidad(cuadratico, tamanos[:5])  # Solo hasta 3000

plt.figure(figsize=(10, 6))
plt.plot(tamanos, t_lineal, 'g-o', label='O(n) - Lineal')
plt.plot(tamanos, t_nlogn, 'b-o', label='O(n log n) - Ordenación')
plt.plot(tamanos[:5], t_cuadratico, 'r-o', label='O(n²) - Cuadrático')
plt.xlabel('Tamaño de datos (n)')
plt.ylabel('Tiempo (segundos)')
plt.title('Comparación de complejidades algorítmicas')
plt.legend()
plt.grid(True)
plt.show()

print("\n¿Qué pasaría con n = 1.000.000?")
print("O(n): ~1 segundo")
print("O(n²): ~11.5 días")
```

## 4. Implicaciones para Big Data

| Complejidad | Con n = 10⁹ datos | Viabilidad |
|-------------|-------------------|------------|
| O(1) | 1 operación | ✅ Ideal |
| O(log n) | ~30 operaciones | ✅ Excelente |
| O(n) | 10⁹ operaciones (~1 seg) | ✅ Viable |
| O(n log n) | 3×10¹⁰ (~30 seg) | ⚠️ Aceptable |
| O(n²) | 10¹⁸ (~31.7 años) | ❌ Imposible |
