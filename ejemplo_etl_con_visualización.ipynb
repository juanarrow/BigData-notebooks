```python
import pandas as pd
url = "https://www.juntadeandalucia.es/datosabiertos/portal/dataset/8555d8b2-5688-4481-bf8c-733020dfc484/resource/300b6ba5-c04d-42c3-8185-687c5d5a8314/download/puntos-de-venta-consorcio-de-transporte-cadiz.csv"
df = pd.read_csv(url, sep=';', encoding='latin1', skiprows=1)
print(df.describe())
print(df.dtypes)
# Create a boolean mask to filter rows with valid lat/long formats
mask = (df['LATITUD'].str.match(r'-?\d{8}$', na=False)) & (df['LONGITUD'].str.match(r'-?\d{7}$', na=False))


# Convert to float and scale correctly on the original dataframe directly
df.loc[mask, 'LATITUD'] = df.loc[mask, 'LATITUD'].astype(float) / 10**6
df.loc[mask, 'LONGITUD'] = df.loc[mask, 'LONGITUD'].astype(float) / 10**6
df.loc[~mask, 'LATITUD'] = None
df.loc[~mask, 'LONGITUD'] = None



import plotly.express as px

# 1. Asegurar que cargamos tus datos limpios filtrados
df_grafica = df[mask].copy()

# 2. Inicializar el mapa interactivo regional con tus columnas
fig = px.scatter_mapbox(
    df_grafica,
    lat="LATITUD",
    lon="LONGITUD",
    hover_name="DIRECCIÓN",
    zoom=9,  # Nivel de zoom perfecto para la provincia de Cádiz
    height=600
)

# 3. Vincular tu API Key de CartoDB en las capas vectoriales del mapa
carto_url_con_key = "https://basemaps.cartocdn.com/rastertiles/voyager/{z}/{x}/{y}.png?key=cb1_4ktg_1_a7212a733336114c72cb7046"

fig.update_layout(
    mapbox=dict(
        style="white-bg",  # Usamos un fondo neutro local para pintar la capa encima
        center=dict(
            lat=df_grafica["LATITUD"].mean(),
            lon=df_grafica["LONGITUD"].mean()
        ),
        layers=[
            {
                "below": 'traces', # Posiciona el mapa por debajo de tus puntos rojos
                "sourcetype": "raster",
                "source": [carto_url_con_key] # Tu URL autorizada
            }
        ]
    ),
    margin={"r":0, "t":0, "l":0, "b":0}
)

# 4. Mostrar el mapa interactivo usando el renderizador oficial de Colab
fig.show(renderer="colab")
```
