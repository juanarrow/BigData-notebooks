## 1. Las 5 Vs del Big Data

| V | Concepto | Descripción |
|---|----------|-------------|
| **Volumen** | Cantidad de datos | Petabytes de datos generados diariamente |
| **Velocidad** | Rapidez de generación | Datos que llegan a alta velocidad |
| **Variedad** | Tipos de datos | Estructurados, semi-estructurados, no estructurados |
| **Veracidad** | Calidad y fiabilidad | Incertidumbre sobre la exactitud |
| **Valor** | Utilidad del dato | La información útil que se puede extraer |

## 2. Notación Big O

La notación Big O describe el **peor caso** de crecimiento de un algoritmo:

| Notación | Nombre | Viable para Big Data |
|----------|--------|---------------------|
| **O(1)** | Constante | ✅ Ideal |
| **O(log n)** | Logarítmica | ✅ Excelente |
| **O(n)** | Lineal | ✅ Bueno |
| **O(n log n)** | Log-lineal | ✅ Aceptable |
| **O(n²)** | Cuadrática | ⚠️ Problemático |
| **O(2ⁿ)** | Exponencial | ❌ Imposible |

## 3. Ejercicio: Medir tiempos de ordenación

```python
import time
import random

# Generar lista de números aleatorios
def generar_lista(n):
    return [random.randint(1, 10000) for _ in range(n)]

# Medir tiempo de ordenación
def medir_tiempo(lista):
    inicio = time.time()
    sorted(lista)  # Ordenación nativa de Python (Timsort)
    fin = time.time()
    return fin - inicio

# Probar con diferentes tamaños
tamanos = [1000, 10000, 100000, 1000000]

print("Tamaño\t\tTiempo (segundos)")
print("-" * 35)

for n in tamanos:
    lista = generar_lista(n)
    tiempo = medir_tiempo(lista)
    print(f"{n:,}\t\t{tiempo:.4f}")
```

## 4. Reflexión

**Pregunta:** ¿El tiempo crece de forma lineal cuando duplicamos los datos?

**Respuesta:** No exactamente. Python usa Timsort que tiene complejidad O(n log n), por lo que al duplicar n, el tiempo se multiplica por aproximadamente 2.x (un poco más que el doble).

## 5. Impacto de la complejidad

Para **n = 1,000,000** registros:

| Complejidad | Operaciones | Tiempo (1μs/op) |
|-------------|-------------|----------------|
| O(1) | 1 | 0.000001 seg |
| O(log n) | 20 | 0.00002 seg |
| O(n) | 1,000,000 | 1 seg |
| O(n log n) | 20,000,000 | 20 seg |
| O(n²) | 1,000,000,000,000 | **11.5 días** |
