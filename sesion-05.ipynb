## 1. Diseño de modelo para E-commerce

````python
import pandas as pd
import numpy as np

# ========================================
# CREAR DATOS DE EJEMPLO (E-COMMERCE)
# ========================================

# Tabla de hechos: Pedidos
pedidos = pd.DataFrame({
    'pedido_id': range(1, 101),
    'cliente_id': [i % 20 + 1 for i in range(100)],
    'fecha': pd.date_range('2024-01-01', periods=100, freq='D'),
    'total': [round(50 + i * 1.5, 2) for i in range(100)]
})

# Líneas de pedido
lineas_pedido = pd.DataFrame({
    'linea_id': range(1, 251),
    'pedido_id': [i % 100 + 1 for i in range(250)],
    'producto_id': [i % 30 + 1 for i in range(250)],
    'cantidad': [1 + i % 5 for i in range(250)],
    'precio_unitario': [10 + (i % 30) * 2 for i in range(250)]
})

# Dimensión: Productos
productos = pd.DataFrame({
    'producto_id': range(1, 31),
    'nombre': [f'Producto_{i}' for i in range(1, 31)],
    'categoria': ['Electrónica', 'Ropa', 'Hogar'] * 10,
    'precio_base': [10 + i * 2 for i in range(30)]
})

# Dimensión: Clientes
clientes = pd.DataFrame({
    'cliente_id': range(1, 21),
    'nombre': [f'Cliente_{i}' for i in range(1, 21)],
    'ciudad': ['Málaga', 'Sevilla', 'Granada', 'Córdoba'] * 5,
    'segmento': ['Premium', 'Standard', 'Básico', 'Premium'] * 5
})

print("=== TABLAS CREADAS ===")
print(f"Pedidos: {len(pedidos)} registros")
print(f"Líneas: {len(lineas_pedido)} registros")
print(f"Productos: {len(productos)} registros")
print(f"Clientes: {len(clientes)} registros")
````

## 2. Análisis dimensional

````python
# ========================================
# ANÁLISIS: Ventas por categoría
# ========================================

# Unir tablas
ventas = lineas_pedido.merge(productos, on='producto_id')
ventas['total_linea'] = ventas['cantidad'] * ventas['precio_unitario']

print("=== VENTAS POR CATEGORÍA ===")
por_categoria = ventas.groupby('categoria').agg({
    'total_linea': 'sum',
    'cantidad': 'sum',
    'linea_id': 'count'
}).rename(columns={'linea_id': 'num_ventas'})

print(por_categoria)
````

````python
# ========================================
# ANÁLISIS: Ventas por cliente y segmento
# ========================================

ventas_cliente = pedidos.merge(clientes, on='cliente_id')

print("=== VENTAS POR SEGMENTO ===")
por_segmento = ventas_cliente.groupby('segmento').agg({
    'total': ['sum', 'mean', 'count']
})
por_segmento.columns = ['total_ventas', 'ticket_medio', 'num_pedidos']

print(por_segmento)
````

````python
# ========================================
# ANÁLISIS TEMPORAL
# ========================================
import matplotlib.pyplot as plt

# Ventas por mes
pedidos['mes'] = pedidos['fecha'].dt.to_period('M')
ventas_mes = pedidos.groupby('mes')['total'].sum()

print("=== VENTAS MENSUALES ===")
print(ventas_mes)

# Gráfico
plt.figure(figsize=(10, 5))
ventas_mes.plot(kind='bar')
plt.title('Ventas por Mes')
plt.xlabel('Mes')
plt.ylabel('Total (€)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
