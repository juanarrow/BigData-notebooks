## Ejercicio 1: Crear KPIs con contexto

```python
import plotly.graph_objects as go

# Datos del negocio
ventas_actual = 125000
ventas_anterior = 108000
objetivo_ventas = 150000

# Crear KPI con contexto
fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number+delta+gauge",
    value=ventas_actual,
    number={'prefix': '€', 'valueformat': ',.0f'},
    delta={'reference': ventas_anterior, 'relative': True},
    title={'text': 'Ventas Mensuales'},
    gauge={
        'axis': {'range': [0, objetivo_ventas * 1.2]},
        'bar': {'color': 'darkblue'},
        'steps': [
            {'range': [0, objetivo_ventas * 0.7], 'color': 'lightcoral'},
            {'range': [objetivo_ventas * 0.7, objetivo_ventas * 0.9], 'color': 'lightyellow'},
            {'range': [objetivo_ventas * 0.9, objetivo_ventas * 1.2], 'color': 'lightgreen'}
        ],
        'threshold': {'value': objetivo_ventas, 'thickness': 0.75}
    }
))

fig.update_layout(height=300)
fig.show()

print(f"\nAnálisis:")
print(f"  Valor actual: €{ventas_actual:,}")
print(f"  vs Anterior: {(ventas_actual/ventas_anterior - 1)*100:+.1f}%")
print(f"  vs Objetivo: {(ventas_actual/objetivo_ventas)*100:.1f}%")
```

## Ejercicio 2: Selección de visualización

**Pregunta:** ¿Qué tipo de gráfico es más adecuado para cada caso?

```python
import pandas as pd
import numpy as np
import plotly.express as px

np.random.seed(42)

# Datos de ejemplo
df = pd.DataFrame({
    'mes': pd.date_range('2024-01', periods=12, freq='ME'),
    'ventas': np.random.randint(80000, 150000, 12),
    'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], 12),
    'categoria': np.random.choice(['A', 'B', 'C'], 12)
})

# Caso 1: Mostrar TENDENCIA temporal
print("=== CASO 1: Tendencia temporal → Gráfico de LÍNEAS ===")
ventas_mes = df.groupby('mes')['ventas'].sum().reset_index()
fig = px.line(ventas_mes, x='mes', y='ventas', markers=True,
              title='Evolución de Ventas (Líneas)')
fig.show()

# Caso 2: COMPARAR categorías
print("\n=== CASO 2: Comparar categorías → Gráfico de BARRAS ===")
ventas_region = df.groupby('region')['ventas'].sum().reset_index()
fig = px.bar(ventas_region, x='region', y='ventas',
             title='Ventas por Región (Barras)')
fig.show()

# Caso 3: Mostrar COMPOSICIÓN
print("\n=== CASO 3: Composición/proporción → Gráfico de TARTA ===")
ventas_cat = df.groupby('categoria')['ventas'].sum()
fig = px.pie(values=ventas_cat.values, names=ventas_cat.index,
             title='Distribución por Categoría (Tarta)', hole=0.3)
fig.show()
```

## Ejercicio 3: Diseño de KPIs SMART

Convierte estos KPIs mal definidos en KPIs SMART:

```python
kpis_mal_definidos = [
    "Mejorar las ventas",
    "Más clientes",
    "Reducir costes"
]

kpis_smart = [
    {
        "original": "Mejorar las ventas",
        "smart": "Aumentar ventas online en España un 15% para Q4 2024",
        "S": "Ventas online en España",
        "M": "+15%",
        "A": "Basado en crecimiento histórico del 10%",
        "R": "Alineado con estrategia de digitalización",
        "T": "Q4 2024"
    },
    {
        "original": "Más clientes",
        "smart": "Captar 500 nuevos clientes premium en el trimestre",
        "S": "Clientes nuevos segmento premium",
        "M": "500 clientes",
        "A": "Promedio actual es 400/trimestre",
        "R": "Segmento de mayor LTV",
        "T": "Un trimestre"
    },
    {
        "original": "Reducir costes",
        "smart": "Reducir coste de adquisición (CAC) de €50 a €40 en 6 meses",
        "S": "Coste de adquisición de cliente (CAC)",
        "M": "De €50 a €40 (-20%)",
        "A": "Optimizando canales digitales",
        "R": "Mejora el margen de contribución",
        "T": "6 meses"
    }
]

print("=== CONVERSIÓN A KPIs SMART ===")
for kpi in kpis_smart:
    print(f"\n❌ Original: {kpi['original']}")
    print(f"✅ SMART: {kpi['smart']}")
    print(f"   S: {kpi['S']}")
    print(f"   M: {kpi['M']}")
    print(f"   A: {kpi['A']}")
    print(f"   R: {kpi['R']}")
    print(f"   T: {kpi['T']}")
```
