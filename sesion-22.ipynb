```python
# ========================================
# RESUMEN DE LIBRERÍAS UTILIZADAS
# ========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

print("=== LIBRERÍAS CARGADAS ===")
print(f"pandas: {pd.__version__}")
print(f"numpy: {np.__version__}")
print("matplotlib, seaborn, plotly, sklearn")
```

```python
# ========================================
# EJEMPLO COMPLETO: Pipeline de datos
# ========================================

# 1. CREAR DATOS
np.random.seed(42)
df = pd.DataFrame({
    'fecha': pd.date_range('2024-01-01', periods=100, freq='D'),
    'ventas': np.random.randint(1000, 5000, 100),
    'categoria': np.random.choice(['A', 'B', 'C'], 100),
    'region': np.random.choice(['Norte', 'Sur', 'Este'], 100)
})

print("1. DATOS CREADOS")
print(df.head())

# 2. TRANSFORMAR
df['mes'] = df['fecha'].dt.to_period('M').astype(str)
resumen = df.groupby(['mes', 'categoria'])['ventas'].sum().reset_index()

print("\n2. DATOS TRANSFORMADOS")
print(resumen.head())

# 3. VISUALIZAR
fig = px.bar(resumen, x='mes', y='ventas', color='categoria',
             title='Ventas por Mes y Categoría', barmode='group')
fig.show()

print("\n3. VISUALIZACIÓN GENERADA ✅")
```
