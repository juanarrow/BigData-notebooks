## 2. Estimación de costes

```python
def estimar_coste_mensual(config):
    """
    Estima el coste mensual de una solución Big Data en AWS
    Precios aproximados de referencia (varían por región)
    """
    costes = {}
    
    # S3 Storage
    s3_gb = config.get('storage_gb', 0)
    costes['S3 Storage'] = s3_gb * 0.023  # ~$0.023/GB/mes
    
    # S3 Requests
    s3_requests_m = config.get('s3_requests_millions', 0)
    costes['S3 Requests'] = s3_requests_m * 0.005  # ~$0.005/1000 requests
    
    # EMR (Spark)
    emr_hours = config.get('emr_hours', 0)
    emr_nodes = config.get('emr_nodes', 1)
    costes['EMR'] = emr_hours * emr_nodes * 0.27  # m5.xlarge
    
    # Athena queries
    athena_tb_scanned = config.get('athena_tb_scanned', 0)
    costes['Athena'] = athena_tb_scanned * 5.0  # $5/TB escaneado
    
    # Data Transfer
    transfer_gb = config.get('data_transfer_gb', 0)
    costes['Data Transfer'] = transfer_gb * 0.09  # ~$0.09/GB
    
    total = sum(costes.values())
    
    return costes, total

# Ejemplo de configuración
config_ejemplo = {
    'storage_gb': 5000,           # 5 TB de datos
    's3_requests_millions': 10,   # 10M requests
    'emr_hours': 200,             # 200 horas de EMR
    'emr_nodes': 5,               # 5 nodos
    'athena_tb_scanned': 2,       # 2 TB escaneados
    'data_transfer_gb': 500       # 500 GB transferidos
}

costes, total = estimar_coste_mensual(config_ejemplo)

print("=== ESTIMACIÓN DE COSTES MENSUALES ===")
print(f"\nConfiguración:")
for k, v in config_ejemplo.items():
    print(f"  {k}: {v}")

print(f"\nDesglose de costes:")
for servicio, coste in costes.items():
    print(f"  {servicio}: ${coste:,.2f}")

print(f"\n{'='*40}")
print(f"TOTAL ESTIMADO: ${total:,.2f}/mes")
print(f"               ${total*12:,.2f}/año")
```

## Ventajas del Cloud para Big Data

1. **Escalabilidad elástica:** Escalar según demanda
2. **Pago por uso:** Sin inversión inicial
3. **Servicios gestionados:** Menor carga operativa
4. **Disponibilidad global:** Despliegue en múltiples regiones
5. **Integración:** Ecosistema completo de servicios
