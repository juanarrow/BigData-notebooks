## 8. Funciones

`def` crea una función. Hay que llamarla con paréntesis: `saluda` es la función y `saluda()` la ejecuta.
`return` devuelve un valor. Si devuelves varios, separados por comas, quien llama puede desempaquetarlos.

Un parámetro puede tener valor por defecto. Si ese valor es una lista o un diccionario, créalo **dentro** de la función: el valor por defecto se construye una sola vez, al definirla.

`lambda` es una función pequeña, de una sola expresión, habitual con `map` y `filter`.


````python
def saludar():
    return "Hola"

def saludar_a(nombre):
    return f"Hola, {nombre}"

def sumar(x, y=10):
    return x + y

print(saludar())
print(saludar_a("Nuria"))
print(sumar(5, 3))
print(sumar(7))


def minimo_y_maximo(numeros):
    return min(numeros), max(numeros)


menor, mayor = minimo_y_maximo([1, 5, 3])
print(menor, mayor)


def anadir(item, caja=None):
    if caja is None:
        caja = []
    caja.append(item)
    return caja


print(anadir("a"))
print(anadir("b"))
````

## 9. Clases

Una clase es el molde. Cada objeto creado con ese molde es una instancia, con sus propios datos.
`__init__` prepara la instancia. `self` es esa instancia: así un método llega a sus datos.

Un atributo escrito en la clase (fuera de `__init__`) es común a todas las instancias.
Un `@classmethod` recibe la clase (`cls`) y sirve para construir objetos con otra receta.
Una clase hija hereda de la padre y puede redefinir un método.


````python
class Medicion:
    def __init__(self, sensor, valor):
        self.sensor = sensor
        self.valor = valor

    def resumen(self):
        return f"{self.sensor}: {self.valor} °C"


lectura = Medicion("aula-1", 21.5)
print(lectura.resumen())


class Sensor:
    unidad = "°C"

    def __init__(self, nombre):
        self.nombre = nombre

    def ficha(self):
        return f"{self.nombre} ({self.unidad})"

    @classmethod
    def desde_codigo(cls, codigo):
        return cls(f"sensor-{codigo}")


aula = Sensor("aula-1")
patio = Sensor.desde_codigo(12)
print(aula.ficha())
print(patio.ficha())
print(aula.unidad, patio.unidad)


class Lectura:
    def __init__(self, valor):
        self.valor = valor

    def etiqueta(self):
        return "lectura"


class LecturaTemperatura(Lectura):
    def etiqueta(self):
        return f"{self.valor} °C"


print(Lectura(21).etiqueta())
print(LecturaTemperatura(21).etiqueta())
````

## 10. Excepciones

Cuando una operación no puede hacerse, Python lanza una excepción y corta el flujo normal.
`try` intenta el código. Cada `except` recoge un tipo de error. `else` se ejecuta si no hubo error. `finally` se ejecuta siempre: sirve para cerrar lo que hayas abierto.

Atrapa el tipo concreto (`ValueError`, `ZeroDivisionError`…), no un `except` vacío.
`raise` lanza una excepción cuando tus datos no cumplen una regla.


````python
for texto in ["10", "abc", "0"]:
    try:
        numero = int(texto)
        resultado = 10 / numero
    except ValueError:
        print(f"{texto!r}: eso no es un número entero")
    except ZeroDivisionError:
        print(f"{texto!r}: no se puede dividir entre cero")
    else:
        print(f"{texto!r}: resultado {resultado}")
    finally:
        print("  intento terminado")
````

````python
ejemplos = [
    ("ValueError", lambda: int("abc")),
    ("TypeError", lambda: "2" + 2),
    ("IndexError", lambda: [1][5]),
    ("KeyError", lambda: {}["clave"]),
    ("FileNotFoundError", lambda: open("este_fichero_no_existe.txt")),
]

for esperado, accion in ejemplos:
    try:
        accion()
    except Exception as error:
        print(f"{esperado}: {type(error).__name__}")


def validar_edad(edad):
    if edad < 0:
        raise ValueError("La edad no puede ser negativa")
    return edad


try:
    validar_edad(-3)
except ValueError as error:
    print(error)
````

## 11. Colecciones

Una colección guarda varios elementos y se puede recorrer con `for`.
Las cadenas también son colecciones de caracteres.
`len()` dice cuántos elementos hay. `in` pregunta si un elemento está dentro.

Cada tipo resuelve una necesidad distinta:

| Tipo | Orden | Duplicados | Para qué |
|---|---|---|---|
| `list` | sí | sí | una secuencia que cambia |
| `tuple` | sí | sí | un grupo fijo, por ejemplo una coordenada |
| `set` | no | no | pertenencia y operaciones de conjuntos |
| `dict` | sí (inserción) | claves únicas | buscar un valor por su clave |


### Listas


````python
vacia = []
numeros = [5]
mezcla = [1, "dos", 3.0, True]
print(vacia, numeros, mezcla)

numeros.append("x")
print("append", numeros)
numeros.insert(0, "y")
print("insert", numeros)
numeros.extend(["z", 5])
print("extend", numeros)
numeros.remove("x")
print("remove", numeros)
ultimo = numeros.pop()
print("pop", ultimo, numeros)

frutas = ["plátano", "manzana", "naranja"]
print(frutas[0], frutas[-1])
print("manzana" in frutas)
print(len(frutas))
````

### Tuplas

Una tupla de un solo elemento lleva coma: `(1,)`. Sin la coma, `(1)` es solo el número 1.
El desempaquetado reparte la tupla en variables. `*resto` se queda con lo que sobra, en una lista.


````python
punto = (3, 4)
una = (1,)
vacia = ()
print(type(punto), type(una), type(vacia))

x, y = punto
print(x, y)

primero, *resto = (1, 2, 3, 4)
print(primero, resto)
````

### Conjuntos

Un conjunto elimina duplicados. `|` es la unión, `&` la intersección, `-` la diferencia y `^` la diferencia simétrica (lo que está en uno o en otro, no en los dos).


````python
a = {1, 2, 3}
b = set([3, 4, 4, 5])
print(b)
print("unión", a | b)
print("intersección", a & b)
print("diferencia", a - b)
print("simétrica", a ^ b)
````

### Diccionarios

La clave va entre corchetes. `.get(clave, defecto)` devuelve el defecto si la clave no está, en lugar de lanzar `KeyError`.
`.keys()`, `.values()` e `.items()` permiten recorrer el diccionario.


````python
vacio = {}
registro = {"ciudad": "Valencia", "temp": 22.1}
print(vacio, registro)

registro["humedad"] = 60
registro["temp"] = 23.0
print(registro.get("temp", 0))
print(registro.get("viento", 0))

del registro["humedad"]
temp = registro.pop("temp")
print(temp, registro)

sensor = {"nombre": "aula-1", "aviso": "calor"}
print(list(sensor.keys()))
print(list(sensor.values()))
print(list(sensor.items()))
````

## 12. Comprensiones

Una comprensión es un bucle escrito en una línea, que construye otra colección.
Lee el ejemplo de izquierda a derecha: *qué* elemento, *de dónde* sale y *con qué* condición.

La de diccionario y la de conjunto cambian los corchetes por llaves.
Una expresión generadora usa paréntesis y no construye la lista entera: calcula cada valor cuando alguien lo pide. Por eso encaja bien dentro de `sum()`.


````python
cuadrados = [x ** 2 for x in range(10)]
pares = [x for x in range(20) if x % 2 == 0]
matriz = [[i * j for j in range(3)] for i in range(3)]
print(cuadrados)
print(pares)
print(matriz)

longitudes = {palabra: len(palabra) for palabra in ["hola", "mundo"]}
unicas = {len(palabra) for palabra in ["qué", "cuál", "por qué"]}
print(longitudes)
print(unicas)

generador = (x ** 2 for x in range(5))
print(generador)
print(sum(x ** 2 for x in range(1000)))
````

## 13. Ficheros

`with open(...)` abre el fichero y lo cierra al salir del bloque, también si hay un error.
Indica `encoding="utf-8"` para que las tildes y la `ñ` se lean igual en cualquier ordenador.

| Modo | Efecto |
|---|---|
| `"r"` | leer |
| `"w"` | escribir, vaciando el fichero si ya existía |
| `"a"` | añadir al final |

`pathlib.Path` construye rutas sin pegar barras a mano. El ejemplo escribe en una carpeta temporal para no dejar ficheros en el repositorio.


````python
import tempfile
from pathlib import Path

carpeta = Path(tempfile.mkdtemp())
ruta = carpeta / "lecturas.txt"
print(ruta.name, ruta.suffix)

with ruta.open(mode="w", encoding="utf-8") as fichero:
    fichero.write("Valencia,22.1\n")
    fichero.write("Teruel,9.4\n")

with ruta.open(mode="r", encoding="utf-8") as fichero:
    contenido = fichero.read()
print("--- fichero completo ---")
print(contenido)

with ruta.open(mode="r", encoding="utf-8") as fichero:
    for linea in fichero:
        print("línea:", linea.strip())

diario = carpeta / "diario.txt"
with diario.open(mode="w", encoding="utf-8") as fichero:
    fichero.write("arranque\n")
with diario.open(mode="a", encoding="utf-8") as fichero:
    fichero.write("nueva entrada\n")
print(diario.read_text(encoding="utf-8"))
````

## 14. Módulos e importaciones

Un módulo es un fichero `.py`. Importar trae sus nombres a tu programa.

- `import math` trae el módulo entero. Usas `math.sqrt`.
- `from math import sqrt` trae solo esa función.
- `import statistics as stats` pone un alias cuando el nombre es largo.

Evita `from modulo import *`: llena tu programa de nombres y no se ve de dónde sale cada uno.

Orden recomendado, con una línea en blanco entre grupos: biblioteca estándar, bibliotecas de terceros, módulos tuyos.


````python
import math

print(math.sqrt(16))

from math import sqrt

print(sqrt(16))

import statistics as stats

print(stats.mean([10, 20, 30]))
````

Un paquete es una carpeta con módulos. La celda siguiente crea uno diminuto en una carpeta temporal y lo importa. Es el mismo mecanismo que usarás cuando el código deje de caber en un solo fichero.

Si editas el fichero y vuelves a ejecutar la celda, reinicia el kernel: Python se queda con la primera versión importada.


````python
import sys
import tempfile
from pathlib import Path

raiz = Path(tempfile.mkdtemp())
(raiz / "paquete").mkdir()
(raiz / "paquete" / "__init__.py").write_text("", encoding="utf-8")
(raiz / "paquete" / "modulo.py").write_text(
    "def saludar(nombre):\n"
    "    return f'Hola, {nombre}'\n"
    "\n"
    "class Sensor:\n"
    "    def __init__(self, nombre):\n"
    "        self.nombre = nombre\n",
    encoding="utf-8",
)
(raiz / "paquete" / "subpaquete").mkdir()
(raiz / "paquete" / "subpaquete" / "__init__.py").write_text("", encoding="utf-8")
(raiz / "paquete" / "subpaquete" / "modulo.py").write_text(
    "def doble(n):\n"
    "    return n * 2\n",
    encoding="utf-8",
)

sys.path.insert(0, str(raiz))

import paquete.modulo
from paquete import modulo
from paquete.subpaquete import modulo as sub
from paquete.modulo import saludar, Sensor
from paquete.modulo import saludar as hola

print(paquete.modulo.saludar("Ana"))
print(modulo.saludar("Luis"))
print(hola("Inés"))
print(sub.doble(21))
print(Sensor("aula-1").nombre)
````

## 17. Valores verdaderos, valores falsos y construcciones habituales

En un `if` o un `while`, Python trata algunos valores como falsos: el número `0`, el decimal `0.0`, las cadenas vacías, las colecciones vacías y `None`. El resto cuenta como verdadero.

Tres patrones que verás a menudo:

- intercambiar dos variables en una línea;
- aplanar una lista de listas;
- quitar duplicados, y contar cuántas veces aparece cada valor con `Counter`.


````python
muestras = [-42, 0, 3.14, 0.0, "Ana", "", [1, 2, 3], [], ("manzana", "pera"), (), {"clave": None}, {}, None]
for valor in muestras:
    print(f"{valor!r:25} -> {bool(valor)}")
````

## 18. Ejemplo integrador

Junta funciones, diccionarios y una comprensión. Es el tipo de transformación que harás con datos reales: agrupar y resumir.


````python
registros = [
    {"ciudad": "Valencia", "temp": 22.1},
    {"ciudad": "Teruel", "temp": 9.4},
    {"ciudad": "Valencia", "temp": 23.0},
    {"ciudad": "Teruel", "temp": 8.1},
    {"ciudad": "Cádiz", "temp": 24.6},
]


def media_por_ciudad(filas):
    grupos = {}
    for fila in filas:
        grupos.setdefault(fila["ciudad"], []).append(fila["temp"])
    return {
        ciudad: round(sum(valores) / len(valores), 2)
        for ciudad, valores in grupos.items()
    }


print(media_por_ciudad(registros))
````
